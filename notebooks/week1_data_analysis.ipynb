```python
import kagglehub

# Download the latest dataset version
path = kagglehub.dataset_download("nehalbirla/vehicle-dataset-from-cardekho")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Locate and load the CSV file
file_path = os.path.join(path, "car data.csv")
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("--- First 5 rows ---")
display(df.head())
```

```python
print("--- DATASET INFO ---")
df.info()
print("\n==================================================\n")
```

```python
print("--- SUMMARY STATISTICS ---")
display(df.describe(include='all'))
```

```python
print("--- INITIAL COUNT OF MISSING VALUES ---")
print(df.isnull().sum())

# Treat zero Kms_Driven values as possible invalid/missing measurements,
# following the approach in the combined internship document.
zero_count = (df["Kms_Driven"] == 0).sum()
if zero_count > 0:
    print(f"Found {zero_count} rows with 0 Kms_Driven. Replacing with missing values and filling with the median.")
    import numpy as np
    df["Kms_Driven"] = df["Kms_Driven"].replace(0, np.nan)
    df["Kms_Driven"] = df["Kms_Driven"].fillna(df["Kms_Driven"].median())
else:
    print("No zero values found in Kms_Driven.")

print("\nMissing values after this step:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()
print(f"Removing {duplicate_count} duplicate rows...")
df = df.drop_duplicates().reset_index(drop=True)

# Standardize categorical text to uppercase
for column in ["Fuel_Type", "Seller_Type", "Transmission"]:
    df[column] = df[column].astype("string").str.upper()

print("Duplicate removal and category formatting completed.")
display(df.head())
```

```python
from datetime import datetime

current_year = datetime.now().year
df["Vehicle_Age"] = current_year - df["Year"]

print("Vehicle_Age feature created using the current year:", current_year)
display(df[['Year', 'Vehicle_Age']].head())
```

```python
print("--- FINAL CLEANED DATASET STRUCTURE ---")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nFirst five cleaned rows:")
display(df.head())

# Save the cleaned dataset in the current notebook working directory
output_file = "cleaned_cardekho_dataset.csv"
df.to_csv(output_file, index=False)
print(f"Cleaned dataset saved as: {output_file}")
```
